## Pergunta 8
Qual gênero favorito apresenta a maior taxa de cancelamento?

```python
taxa_churn_genero = (
    df.groupby("favorite_genre")["churned"]
      .mean()
      .mul(100)
      .sort_values(ascending=False)
)

taxa_churn_genero
```

## Pergunta 10
Clientes que assistem mais horas apresentam menor taxa de cancelamento?

```python
df["faixa_watch"] = pd.cut(
    df["watch_hours"],
    bins=[0, 20, 40, 60, float("inf")],
    labels=["Até 20h", "21-40h", "41-60h", "Mais de 60h"]
)
```

```python
df.groupby("faixa_watch", observed=True)["churned"].mean().mul(100)
```
